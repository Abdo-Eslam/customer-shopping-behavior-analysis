```python
# Import required libraries
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
# Display first 5 rows to understand the structure
df.head()
```

```python
# Check dataset structure and data types
df.info()
```

```python
# Summary statistics for both numerical and categorical columns
df.describe(include='all')
```

```python
# Check missing values in each column
df.isnull().sum()
```

```python
# Fill missing review ratings using category-level median
# Median helps reduce the impact of outliers
df['Review Rating'] = df.groupby('Category')['Review Rating'] \
                         .transform(lambda x: x.fillna(x.median()))
```

```python
# Validate that no missing values remain
assert df['Review Rating'].isnull().sum() == 0
```

```python
# Convert column names to lowercase
df.columns = df.columns.str.lower()
```

```python
# Replace spaces with underscores
df.columns = df.columns.str.replace(' ', '_')
```

```python
# Rename purchase_amount_(usd) for simplicity
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
# Create age groups using quartiles
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)

df[['age','age_group']].head()
```

```python
# Map purchase frequency to number of days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
# Check if discount_applied and promo_code_used contain identical values
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Drop redundant column
df = df.drop('promo_code_used', axis=1)
```

## 🔗 Database Integration (Microsoft SQL Server)

The cleaned dataset is exported to Microsoft SQL Server to support structured SQL-based business analysis.
The connection is established using Windows Authentication to ensure secure integration within a local development environment.

### 📦 Environment Setup

All required Python libraries are listed in `requirements.txt`.

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# SQL Server connection details
server = "ABDO_ESLAM"
database = "customer_behavior"

# Use Windows Authentication
driver = quote_plus("ODBC Driver 17 for SQL Server")

engine = create_engine(
    f"mssql+pyodbc://@{server}/{database}?driver={driver}&trusted_connection=yes"
)

# Export DataFrame to SQL Server
df.to_sql("customer", engine, if_exists="replace", index=False)

print("Data successfully loaded into Microsoft SQL Server.")
```
